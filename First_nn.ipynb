```python
import numpy as np
import pandas as pd
import torch.nn as nn
import torch.nn.functional as F
```

```python
class Model(nn.Module):
    def __init__(self , input_params = 4 , h1 = 8 , h2 = 9 ,  output_params = 3):
      super().__init__()
      self.fc1 = nn.Linear(input_params , h1)
      self.fc2 = nn.Linear(h1, h2)
      self.out = nn.Linear(h2, output_params)
    def forward(self , x):
      x = F.relu(self.fc1(x))
      x = F.relu(self.fc2(x))
      x = self.out(x)
      return x


```
